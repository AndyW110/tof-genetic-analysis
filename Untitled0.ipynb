```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
NAVY,RED,STEEL,GOLD="#17324d","#a2353a","#5b7c99","#c8963e"
plt.rcParams.update({"font.size":11,"axes.grid":True,"grid.color":"#e6e9ec","figure.dpi":120})
t1=pd.read_excel("tof_supp/Table1.xlsx")
t1["CADD"]=pd.to_numeric(t1["CADD"],errors="coerce")
t1["Position"]=pd.to_numeric(t1["Position"],errors="coerce")
t1["Gene"]=t1["Gene"].str.strip().str.upper()
t1["FetalExpr"]=pd.to_numeric(t1["Fetal Heart Expression"],errors="coerce")
LOF={"stop gained","frameshift variant","stop lost","start loss","start lost"}
t1["is_LoF"]=t1["Mutation"].str.strip().isin(LOF)
print("variants:",len(t1),"| probands:",t1["Proband Study ID"].nunique(),
      "| genes:",t1["Gene"].nunique(),"| CADD>=20:",int((t1["CADD"]>=20).sum()),
      "| LoF:",int(t1["is_LoF"].sum()))
s5=pd.read_excel("supplementary_tables_chd_wgs.xlsx",sheet_name="S5 CHD gene lists",header=1)
def cg(i): return set(s5.iloc[:,i].dropna().astype(str).str.strip().str.upper())-{""}
human_chd,chromatin,cilia=cg(1),cg(3),cg(4)
t1["in_hCHD"]=t1["Gene"].isin(human_chd); t1["in_chrom"]=t1["Gene"].isin(chromatin); t1["in_cilia"]=t1["Gene"].isin(cilia)
t3raw=pd.read_excel("tof_supp/Table3.xlsx",header=None)
page=set(g.strip().upper() for g in str(t3raw.iloc[2,4]).split(",")); t1["in_page"]=t1["Gene"].isin(page)

cons=t1["Mutation"].str.strip().value_counts()
fig,ax=plt.subplots(figsize=(7.2,3.8))
ax.barh(cons.index[::-1],cons.values[::-1],color=[RED if c in LOF else STEEL for c in cons.index][::-1])
for i,v in enumerate(cons.values[::-1]): ax.text(v+1.5,i,str(v),va="center",fontsize=10,color=NAVY)
ax.set_title("Consequence classes of 183 TOF de novo variants",color=NAVY)
ax.set_xlabel("number of variants"); ax.set_xlim(0,cons.max()*1.12)
plt.tight_layout(); plt.savefig("fig1_consequence.png",bbox_inches="tight"); plt.show()

fig,ax=plt.subplots(1,2,figsize=(8.6,3.6))
ax[0].pie([t1["is_LoF"].sum(),(~t1["is_LoF"]).sum()],labels=["LoF (n=%d)"%t1["is_LoF"].sum(),"damaging missense/\ninframe (n=%d)"%(~t1["is_LoF"]).sum()],colors=[RED,STEEL],startangle=90,wedgeprops={"width":0.42},textprops={"fontsize":10}); ax[0].set_title("All 183 variants",color=NAVY)
hi=t1[t1["CADD"]>=20]
ax[1].pie([hi["is_LoF"].sum(),(~hi["is_LoF"]).sum()],labels=["LoF (n=%d)"%hi["is_LoF"].sum(),"damaging missense/\ninframe (n=%d)"%(~hi["is_LoF"]).sum()],colors=[RED,STEEL],startangle=90,wedgeprops={"width":0.42},textprops={"fontsize":10}); ax[1].set_title("CADD ≥ 20 subset (n=%d)"%len(hi),color=NAVY)
plt.tight_layout(); plt.savefig("fig2_lof_donuts.png",bbox_inches="tight"); plt.show()

fig,ax=plt.subplots(figsize=(7.4,3.8))
ax.hist(t1["CADD"].dropna(),bins=np.arange(0,52,2),color=STEEL,edgecolor="white")
ax.axvline(20,color=RED,lw=1.6,ls="--"); ax.text(33.5,ax.get_ylim()[1]*0.9,"threshold CADD = 20",color=RED,fontsize=10)
ax.set_title("CADD score distribution of TOF de novo variants",color=NAVY); ax.set_xlabel("CADD score"); ax.set_ylabel("number of variants")
plt.tight_layout(); plt.savefig("fig3_cadd_dist.png",bbox_inches="tight"); plt.show()

lab=["CHD7\n(known TOF)","WASHC5\n(known TOF)","CLUH\n(novel candidate)","UNC13C\n(novel candidate)"]
fig,ax=plt.subplots(figsize=(7.2,3.4))
ax.bar(range(4),[2,2,2,2],color=[GOLD,GOLD,NAVY,NAVY])
for i in range(4): ax.text(i,2.03,"2",ha="center",fontsize=10,color=NAVY)
ax.set_xticks(range(4)); ax.set_xticklabels(lab,fontsize=9.5); ax.set_ylim(0,2.4)
ax.set_title("Genes hit by de novo variants in ≥ 2 TOF probands",color=NAVY); ax.set_ylabel("number of probands")
plt.tight_layout(); plt.savefig("fig4_recurrent.png",bbox_inches="tight"); plt.show()

sets={"Known TOF/CHD genes (Page 2023, 13)":t1["in_page"].sum(),"Chromatin genes":t1["in_chrom"].sum(),"Cilia genes":t1["in_cilia"].sum(),"Human CHD genes (Richter 2020)":t1["in_hCHD"].sum()}
fig,ax=plt.subplots(figsize=(7.6,3.9))
ax.barh(list(sets)[::-1],list(sets.values())[::-1],color=[GOLD,NAVY,RED,STEEL][::-1])
for i,v in enumerate(list(sets.values())[::-1]): ax.text(v+1.2,i,"%d (%.1f%%)"%(v,100*v/len(t1)),va="center",fontsize=10,color=NAVY)
ax.set_xlim(0,max(sets.values())*1.3); ax.set_title("TOF variants overlapping curated gene sets",color=NAVY); ax.set_xlabel("number of variants")
plt.tight_layout(); plt.savefig("fig5_genesets.png",bbox_inches="tight"); plt.show()

t3c=t3raw.iloc[2:].copy(); t3c.columns=["Categories","Term","p","BH","Molecules","n"]
t3c["BH"]=pd.to_numeric(t3c["BH"],errors="coerce"); t3c["n"]=pd.to_numeric(t3c["n"],errors="coerce")
curated=["Congenital anomaly of cardiovascular system","Morphogenesis of cardiovascular system","Morphogenesis of blood vessel","Formation of heart ventricle","Development of heart septum","Muscle contraction","Organization of muscle cells","Organization of cytoskeleton"]
t3c=t3c.dropna(subset=["BH"]).sort_values("BH").drop_duplicates("Term")
top=t3c[t3c["Term"].isin(curated)].sort_values("BH").iloc[::-1]
fig,ax=plt.subplots(figsize=(9.0,4.6))
bars=ax.barh(top["Term"],-np.log10(top["BH"]),color=STEEL)
for rect,n in zip(bars,top["n"]): ax.text(rect.get_width()+0.05,rect.get_y()+rect.get_height()/2,"n=%d"%int(n),va="center",fontsize=9,color=NAVY)
ax.set_title("Cardiovascular terms enriched among CADD≥20 TOF variants",color=NAVY); ax.set_xlabel("−log10(B–H p-value)")
ax.set_xlim(0,(-np.log10(top["BH"])).max()*1.14)
plt.tight_layout(); plt.savefig("fig6_pathways.png",bbox_inches="tight"); plt.show()

ct=pd.crosstab(t1["is_LoF"],t1["in_page"]); odr,pf=stats.fisher_exact(ct.values)
a=t1.loc[t1["in_page"],"CADD"].dropna(); b=t1.loc[~t1["in_page"],"CADD"].dropna()
u,pm=stats.mannwhitneyu(a,b,alternative="two-sided"); ex=t1["FetalExpr"].dropna()
print("Fisher OR=%.3f p=%.4f"%(odr,pf))
print("CADD known-TOF median=%.1f vs other=%.1f, Mann-Whitney p=%.4f"%(a.median(),b.median(),pm))
print("fetal-heart expressed: %d/%d (%.1f%%)"%((ex>0).sum(),len(ex),100*(ex>0).mean()))
t1.to_csv("tof_variants_clean.csv",index=False); print("ALL DONE")
```
